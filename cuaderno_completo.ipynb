```python
import json
import sys

# Lista de notebooks en el orden deseado
archivos = [
    'introducciónpython.ipynb',
    'PAO25_25_02.ipynb',
    'PAO25_25_03.ipynb',
    'PAO25_25_04.ipynb',
    'PAO25_25_05_Python_Files.ipynb',
    'PAO25_25_06.ipynb',
    'PAO25_25_07.ipynb',
    'PAO25_25_08_Python_Pandas.ipynb',
    'PAO25_25_09_Python_Visualization.ipynb',
    'repaso.ipynb'
]

celdas_combinadas = []

for archivo in archivos:
    try:
        with open(archivo, 'r', encoding='utf-8') as f:
            notebook = json.load(f)
            celdas_combinadas.extend(notebook['cells'])
    except Exception as e:
        print(f"Error al procesar {archivo}: {e}")
        sys.exit(1)

# Crear el notebook combinado (usando la estructura del primer notebook)
with open(archivos[0], 'r', encoding='utf-8') as f:
    plantilla = json.load(f)

plantilla['cells'] = celdas_combinadas

# Guardar el resultado
with open('Curso_Completo.ipynb', 'w', encoding='utf-8') as f:
    json.dump(plantilla, f, ensure_ascii=False, indent=1)

print("¡Notebooks combinados exitosamente en 'Curso_Completo.ipynb'!")
```
